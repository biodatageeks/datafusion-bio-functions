# Lance Encoding Sandbox

Runs the Rust sandbox CLI and renders JSON report tables. Timing is produced by Rust, not Python.

```python
from pathlib import Path
import json, subprocess
import pandas as pd

repo = Path('/Users/mwiewior/research/git/datafusion-bio-functions')
crate = repo / 'research/lance_encoding_sandbox/crates/lance_sandbox/Cargo.toml'
config = repo / 'research/lance_encoding_sandbox/configs/current_v21_zstd3.toml'
cmd = ['cargo', 'run', '--release', '--manifest-path', str(crate), '--', 'run', '--config', str(config)]
subprocess.run(cmd, check=True, cwd=repo)
```
